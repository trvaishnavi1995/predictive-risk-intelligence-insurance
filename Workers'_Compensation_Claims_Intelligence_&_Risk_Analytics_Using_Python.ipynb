**Accident Year**

The Accident Date field was transformed to extract the accident year for trend analysis.

**Insights**

Enables yearly claims trend analysis.

Useful for monitoring insurance exposure growth.

```python
df_clean['accident_month'] = (
    df_clean['Accident Date']
    .dt.month_name()
)
```

**Accident Month**

The Accident Month field was transformed to extract the accident year for trend analysis.

**Insights**

Enables monthly claims trend analysis.
Supports identification of claim frequency changes over time.
Useful for monitoring insurance exposure growth.

```python
df_clean['accident_quarter'] = (
    df_clean['Accident Date']
    .dt.quarter
)
```

**Accident Quarter**

Quarter information was extracted from accident dates to identify seasonal injury patterns.

**Insights**

Seasonal claim patterns can be identified.
Certain industries may experience higher injury frequency during specific quarters.
Supports workforce and safety planning.

```python
df_clean['Age Group'] = pd.cut(
    df_clean['Age at Injury'],
    bins=[0,25,35,45,55,65,100],
    labels=[
        '18-25',
        '26-35',
        '36-45',
        '46-55',
        '56-65',
        '65+'
    ]
)

df_clean['Age Group'].value_counts()
```

**Age Group Segmentation**

Age at Injury was categorized into meaningful age groups to simplify demographic analysis.

**Insights**

Identifies vulnerable workforce segments.
Supports workforce risk profiling.
Useful for safety and injury prevention strategies.

```python
df_clean['Claim Severity'] = df_clean['Claim Injury Type'].replace({
    '2. NON-COMP':'Low',
    '3. MED ONLY':'Low',
    '4. TEMPORARY':'Medium',
    '5. PPD SCH LOSS':'High',
    '6. PPD NSL':'High',
    '7. PTD':'Critical',
    '8. DEATH':'Critical'
})
```

**Claim Severity Categorization**

Claim Injury Types were grouped into severity levels ranging from Low Risk to Critical Risk.

**Insights**

Simplifies claim outcome analysis.
Enables risk segmentation.
Supports prioritization of high-impact claims.

```python
df_clean['Claim Processing Days'] = (
    df_clean['ANCR Date'] -
    df_clean['Assembly Date']
).dt.days
```

**Claim Processing Days**

A new feature was created to calculate the number of days between claim assembly and ANCR determination.

**Insights**

Measures claim handling efficiency.
Longer processing times may indicate complex claims.
Helps identify operational bottlenecks.

```python
median_wage = df_clean['Average Weekly Wage (AWW)'].median()

df_clean['Wage Category'] = np.where(
    df_clean['Average Weekly Wage (AWW)'] >= median_wage,
    'High Wage',
    'Low Wage'
)
```

**Wage Category**

Average Weekly Wage values were segmented into High Wage and Low Wage categories using the median wage.

**Insights**

Supports demographic and economic analysis.
Enables comparison of claim patterns across income groups.
Useful for understanding workforce exposure.

```python
df_clean['Risk Score'] = (
    df_clean['Hearing Count'].fillna(0) * 2
    +
    df_clean['Closed Count'].fillna(0)
)
```

**Risk Score Development**

A simple risk score was developed using hearing count and claim closure metrics.

**Insights**

Provides a quantitative measure of claim complexity.
Supports risk ranking.
Enables comparative analysis across claim groups.

Outliers were examined using boxplots and the Interquartile Range (IQR) method. Particular attention was given to wage-related variables.

```python
plt.figure(figsize=(8,4))

sns.boxplot(
    x=df_clean['Average Weekly Wage (AWW)']
)

plt.show()
```

IQR Method

```python
Q1 = df_clean['Average Weekly Wage (AWW)'].quantile(0.25)

Q3 = df_clean['Average Weekly Wage (AWW)'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
```

```python
df_clean = df_clean[
    (df_clean['Average Weekly Wage (AWW)'] >= lower)
    &
    (df_clean['Average Weekly Wage (AWW)'] <= upper)
]
```

**Based on historical claim trends, what is the expected future claim volume?**

```python
# Claims trend by year

yearly_claims = (
    df_clean.groupby('Accident Year')
    .size()
    .reset_index(name='Claim Count')
)

# Moving Average Forecast

yearly_claims['Forecast'] = (
    yearly_claims['Claim Count']
    .rolling(window=3)
    .mean()
)

plt.figure(figsize=(12,6))

plt.plot(
    yearly_claims['Accident Year'],
    yearly_claims['Claim Count'],
    marker='o',
    label='Actual Claims'
)

plt.plot(
    yearly_claims['Accident Year'],
    yearly_claims['Forecast'],
    marker='s',
    linestyle='--',
    label='Forecast Trend'
)

plt.title('Predictive Analysis: Claims Trend Forecast')
plt.xlabel('Year')
plt.ylabel('Number of Claims')
plt.legend()

plt.show()
```

**Interpretation**

**Features Used**

Accident Year

Claim Count

**Insights**
